```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = 'https://raw.githubusercontent.com/kkathait/D501_Final_Project/refs/heads/main/combined_data.csv'

df_premier = pd.read_csv(url)
```

```python
teams = sorted(set(df_premier["HomeTeam"]) | set(df_premier["AwayTeam"]))

records = []

for team in teams:
    home_games = df_premier[df_premier["HomeTeam"] == team]
    
    home_wins = (home_games["FTR"] == "H").sum()
    home_losses = (home_games["FTR"] == "A").sum()
    home_draws = (home_games["FTR"] == "D").sum()
    
    away_games = df_premier[df_premier["AwayTeam"] == team]
    
    away_wins = (away_games["FTR"] == "A").sum()
    away_losses = (away_games["FTR"] == "H").sum()
    away_draws = (away_games["FTR"] == "D").sum()
    
    records.append({
        "Team": team,
        "Wins": home_wins + away_wins,
        "Losses": home_losses + away_losses,
        "Draws": home_draws + away_draws,
        "Games": home_wins + away_wins + home_losses + away_losses + home_draws + away_draws
    })

standings = pd.DataFrame(records)

standings = standings.sort_values("Games", ascending = False)

standings
```

```python
standings["Points"] = (3 * standings["Wins"]) + standings["Draws"]

standings["Games"] = (standings["Wins"] + standings["Losses"] + standings["Draws"])

standings = standings.sort_values("Games", ascending = False)

standings
```

```python
#Calculate cards by referee

referee_cards = (
    df_premier
    .groupby("Referee")
    .agg(
        Home_Yellow_Cards=("HY", "sum"),
        Away_Yellow_Cards=("AY", "sum"),
        Home_Red_Cards=("HR", "sum"),
        Away_Red_Cards=("AR", "sum"),
        Games_Refereed=("Referee", "count")
    )
    .reset_index()
)

referee_cards
```

```python
#Calculate the average cards per game
referee_cards["Home_Yellow_Avg"] = (
    referee_cards["Home_Yellow_Cards"] /
    referee_cards["Games_Refereed"]
)

referee_cards["Away_Yellow_Avg"] = (
    referee_cards["Away_Yellow_Cards"] /
    referee_cards["Games_Refereed"]
)

referee_cards["Home_Red_Avg"] = (
    referee_cards["Home_Red_Cards"] /
    referee_cards["Games_Refereed"]
)

referee_cards["Away_Red_Avg"] = (
    referee_cards["Away_Red_Cards"] /
    referee_cards["Games_Refereed"]
)

referee_cards
```

```python
#Compare home vs. away cards
referee_cards["Yellow_Card_Difference"] = (
    referee_cards["Home_Yellow_Avg"] -
    referee_cards["Away_Yellow_Avg"]
)

referee_cards["Red_Card_Difference"] = (
    referee_cards["Home_Red_Avg"] -
    referee_cards["Away_Red_Avg"]
)

referee_cards.sort_values(
    "Yellow_Card_Difference",
    ascending=False
)
```

```python
#Find the referees with the biggest differences
referee_cards[
    [
        "Referee",
        "Games_Refereed",
        "Home_Yellow_Avg",
        "Away_Yellow_Avg",
        "Home_Red_Avg",
        "Away_Red_Avg",
        "Yellow_Card_Difference",
        "Red_Card_Difference"
    ]
].sort_values(
    "Yellow_Card_Difference",
    ascending=False
)
```

```python
yellow_correlation = referee_cards[
    ["Home_Yellow_Avg", "Away_Yellow_Avg"]
].corr()

red_correlation = referee_cards[
    ["Home_Red_Avg", "Away_Red_Avg"]
].corr()

print("Yellow Card Correlation:")
print(yellow_correlation)

print("\nRed Card Correlation:")
print(red_correlation)
```

```python
import plotly.express as px

plot_data = referee_cards.melt(
    id_vars="Referee",
    value_vars=[
        "Home_Yellow_Avg",
        "Away_Yellow_Avg"
    ],
    var_name="Team_Location",
    value_name="Average_Yellow_Cards"
)

fig = px.bar(
    plot_data,
    x="Referee",
    y="Average_Yellow_Cards",
    color="Team_Location",
    barmode="group",
    title="Average Yellow Cards by Referee: Home vs. Away Teams"
)

fig.show()
```

```python
#Is there a correlation between the number of cards a referee gives to the home team versus the away team?
referee_cards = (
    df_premier
    .groupby("Referee")
    .agg(
        Home_Yellow=("HY", "mean"),
        Away_Yellow=("AY", "mean"),
        Home_Red=("HR", "mean"),
        Away_Red=("AR", "mean")
    )
    .reset_index()
)

# Correlation between home and away yellow cards
yellow_corr = referee_cards["Home_Yellow"].corr(
    referee_cards["Away_Yellow"]
)

# Correlation between home and away red cards
red_corr = referee_cards["Home_Red"].corr(
    referee_cards["Away_Red"]
)

print("Yellow Card Correlation:", yellow_corr)
print("Red Card Correlation:", red_corr)
```

# Final Analysis

```python
import pandas as pd
import numpy as np
import plotly.express as px

url = 'https://raw.githubusercontent.com/kkathait/D501_Final_Project/refs/heads/main/combined_data.csv'

df_premier = pd.read_csv(url)

# Goals
df_premier["Total_Goals"] = df_premier["FTHG"] + df_premier["FTAG"]

# Cards
df_premier["Home_Cards"] = df_premier["HY"] + df_premier["HR"]
df_premier["Away_Cards"] = df_premier["AY"] + df_premier["AR"]

df_premier["Total_Yellow_Cards"] = df_premier["HY"] + df_premier["AY"]
df_premier["Total_Red_Cards"] = df_premier["HR"] + df_premier["AR"]
df_premier["Total_Cards"] = (
    df_premier["Total_Yellow_Cards"] +
    df_premier["Total_Red_Cards"]
)

# Numerical result
df_premier["Home_Win"] = (df_premier["FTR"] == "H").astype(int)
df_premier["Away_Win"] = (df_premier["FTR"] == "A").astype(int)
df_premier["Draw"] = (df_premier["FTR"] == "D").astype(int)

df_premier.head()
```

```python
# Which referees give the most cards?
referee_analysis = (
    df_premier
    .groupby("Referee")
    .agg(
        Games=("Referee", "count"),
        Avg_Goals=("Total_Goals", "mean"),
        Avg_Yellow_Cards=("Total_Yellow_Cards", "mean"),
        Avg_Red_Cards=("Total_Red_Cards", "mean"),
        Avg_Total_Cards=("Total_Cards", "mean"),
        Home_Win_Rate=("Home_Win", "mean"),
        Away_Win_Rate=("Away_Win", "mean"),
        Draw_Rate=("Draw", "mean")
    )
    .reset_index()
)

referee_analysis.sort_values(
    "Avg_Total_Cards",
    ascending=False
)
```

```python
# Does the referee have an association with goals?
correlation_table = referee_analysis[
    [
        "Avg_Goals",
        "Avg_Yellow_Cards",
        "Avg_Red_Cards",
        "Avg_Total_Cards",
        "Home_Win_Rate",
        "Away_Win_Rate"
    ]
].corr()

correlation_table
```

```python
print(
    "Correlation between total cards and goals:",
    referee_analysis["Avg_Total_Cards"].corr(
        referee_analysis["Avg_Goals"]
    )
)

print(
    "Correlation between yellow cards and goals:",
    referee_analysis["Avg_Yellow_Cards"].corr(
        referee_analysis["Avg_Goals"]
    )
)

print(
    "Correlation between red cards and goals:",
    referee_analysis["Avg_Red_Cards"].corr(
        referee_analysis["Avg_Goals"]
    )
)
```

```python
# Does the referee affect home-team winning?
home_win_analysis = (
    df_premier
    .groupby("Referee")
    .agg(
        Games=("Referee", "count"),
        Home_Wins=("Home_Win", "sum"),
        Away_Wins=("Away_Win", "sum"),
        Draws=("Draw", "sum")
    )
    .reset_index()
)

home_win_analysis["Home_Win_Rate"] = (
    home_win_analysis["Home_Wins"] /
    home_win_analysis["Games"]
)

home_win_analysis["Away_Win_Rate"] = (
    home_win_analysis["Away_Wins"] /
    home_win_analysis["Games"]
)

home_win_analysis["Draw_Rate"] = (
    home_win_analysis["Draws"] /
    home_win_analysis["Games"]
)

home_win_analysis.sort_values(
    "Home_Win_Rate",
    ascending=False
)
```

```python
print(
    "Referee average goals vs home win rate:",
    referee_analysis["Avg_Goals"].corr(
        referee_analysis["Home_Win_Rate"]
    )
)

print(
    "Referee average cards vs home win rate:",
    referee_analysis["Avg_Total_Cards"].corr(
        referee_analysis["Home_Win_Rate"]
    )
)
```

```python
# Does the referee give more cards to the losing team?
def losing_team_cards(row):
    if row["FTR"] == "H":
        return row["Away_Cards"]
    elif row["FTR"] == "A":
        return row["Home_Cards"]
    else:
        return np.nan

def winning_team_cards(row):
    if row["FTR"] == "H":
        return row["Home_Cards"]
    elif row["FTR"] == "A":
        return row["Away_Cards"]
    else:
        return np.nan

df_premier["Winning_Team_Cards"] = df_premier.apply(
    winning_team_cards,
    axis=1
)

df_premier["Losing_Team_Cards"] = df_premier.apply(
    losing_team_cards,
    axis=1
)
```

```python
print(
    "Average cards for winning team:",
    df_premier["Winning_Team_Cards"].mean()
)

print(
    "Average cards for losing team:",
    df_premier["Losing_Team_Cards"].mean()
)
```

```python
# Does referee identity predict the match result?
fig = px.bar(
    home_win_analysis.sort_values("Home_Win_Rate", ascending=False),
    x="Referee",
    y="Home_Win_Rate",
    title="Home-Team Win Rate by Referee",
    labels={
        "Home_Win_Rate": "Home Win Rate"
    }
)

fig.show()
```

```python
fig = px.bar(
    home_win_analysis.sort_values("Away_Win_Rate", ascending=False),
    x="Referee",
    y="Away_Win_Rate",
    title="Away-Team Win Rate by Referee",
    labels={
        "Away_Win_Rate": "Away Win Rate"
    }
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

x = referee_analysis["Avg_Total_Cards"]
y = referee_analysis["Avg_Goals"]

# Calculate trendline
slope, intercept = np.polyfit(x, y, 1)

# Predicted values
y_pred = slope * x + intercept

fig = go.Figure()

# Scatter points
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode="markers",
    text=referee_analysis["Referee"],
    marker=dict(
        size=referee_analysis["Games"] / 5
    ),
    hovertemplate=(
        "<b>%{text}</b><br>"
        "Average Cards: %{x:.2f}<br>"
        "Average Goals: %{y:.2f}"
        "<extra></extra>"
    ),
    name="Referees"
))

# Trendline
fig.add_trace(go.Scatter(
    x=x,
    y=y_pred,
    mode="lines",
    name="Trend Line"
))

fig.update_layout(
    title="Referee Cards vs. Average Goals Scored",
    xaxis_title="Average Total Cards per Match",
    yaxis_title="Average Goals per Match"
)

fig.show()
```

```python
correlation = referee_analysis["Avg_Total_Cards"].corr(
    referee_analysis["Avg_Goals"]
)

print("Correlation between referee cards and goals:", correlation)
```

```python
r_squared = correlation ** 2

print("Correlation:", correlation)
print("R²:", r_squared)
```
