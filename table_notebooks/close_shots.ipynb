```python
import pandas as pd
import numpy as np
```

```python
df_premier = pd.read_csv("../combined_data.csv")
```

```python
teams = sorted(set(df_premier["HomeTeam"]) | set(df_premier["AwayTeam"]))

records = []

for team in teams:
    # Home games
    home_games = df_premier[df_premier["HomeTeam"] == team]
    
    home_sot = home_games["HST"].sum()
    home_shots_for = home_games["HS"].sum()

    # Away games
    away_games = df_premier[df_premier["AwayTeam"] == team]
    
    away_sot = away_games["AST"].sum()
    away_shots_for = away_games["AS"].sum()
    
    records.append({
        "Team": team,
        "Games": len(home_games) + len(away_games),
        "Close Shots Home": home_sot,
        "Percent Close Shots Home": home_sot / home_shots_for if home_shots_for > 0 else 0,
        "Close Shots Away": away_sot,
        "Percent Close Shots Away": away_sot / away_shots_for if away_shots_for > 0 else 0,
    })

close_shots_table = pd.DataFrame(records)

close_shots_table["Percent Close Shots Home"] = close_shots_table["Percent Close Shots Home"].round(2)
close_shots_table["Percent Close Shots Away"] = close_shots_table["Percent Close Shots Away"].round(2)

close_shots_table = close_shots_table.sort_values(
    by="Percent Close Shots Home",
    ascending=False).reset_index(drop=True)

close_shots_table
```

```python
teams = sorted(set(df_premier["HomeTeam"]) | set(df_premier["AwayTeam"]))

records = []

for team in teams:
    # Home games
    home_games = df_premier[df_premier["HomeTeam"] == team]
    
    home_sot_against = home_games["AST"].sum()
    home_shots_against = home_games["AS"].sum()

    # Away games
    away_games = df_premier[df_premier["AwayTeam"] == team]
    
    away_sot_against = away_games["HST"].sum()
    away_shots_against = away_games["HS"].sum()
    
    records.append({
        "Team": team,
        "Games": len(home_games) + len(away_games),
        "Close Shots Allowed Home": home_sot_against,
        "Percent Close Shots Allowed Home": home_sot_against / home_shots_against if home_shots_against > 0 else 0,
        "Close Shots Allowed Away": away_sot_against,
        "Percent Close Shots Allowed Away": away_sot_against / away_shots_against if away_shots_against > 0 else 0,
    })

close_shots_allowed_table = pd.DataFrame(records)

close_shots_allowed_table["Percent Close Shots Allowed Home"] = close_shots_allowed_table["Percent Close Shots Allowed Home"].round(2)
close_shots_allowed_table["Percent Close Shots Allowed Away"] = close_shots_allowed_table["Percent Close Shots Allowed Away"].round(2)

close_shots_allowed_table = close_shots_allowed_table.sort_values(
    by="Percent Close Shots Allowed Home",
    ascending=True).reset_index(drop=True)

close_shots_allowed_table
```
