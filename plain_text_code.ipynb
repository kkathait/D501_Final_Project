```python
import pandas as pd
df = pd.read_csv("combined_data.csv")
```

```python

goals_by_team = pd.concat([
    df.groupby("HomeTeam")["FTHG"].sum().rename_axis("team"),
    df.groupby("AwayTeam")["FTAG"].sum().rename_axis("team")
]).groupby(level=0).sum()

most_goals = goals_by_team[goals_by_team == goals_by_team.max()]

print("Team(s) with the most goals:")
print(most_goals.to_string())


# Total wins by each team, whether the win was home or away
home_wins = df.loc[df["FTR"] == "H", "HomeTeam"].value_counts()
away_wins = df.loc[df["FTR"] == "A", "AwayTeam"].value_counts()
wins_by_team = home_wins.add(away_wins, fill_value=0)
most_wins = wins_by_team[wins_by_team == wins_by_team.max()]

print("\nTeam(s) with the most wins:")
print(most_wins.to_string())
```

```python
home_goals = (
    df.groupby("HomeTeam")["FTHG"]
    .mean()
    .rename("avg_goals_at_home")
)

away_goals = (
    df.groupby("AwayTeam")["FTAG"]
    .mean()
    .rename("avg_goals_away")
)

team_goal_averages = pd.concat([home_goals, away_goals], axis=1).round(2)

print(team_goal_averages.to_string())
```

```python
home_team = "Man City"
away_team = "Arsenal"

print(f"{home_team} average goals at home: "
      f"{team_goal_averages.loc[home_team, 'avg_goals_at_home']}")

print(f"{away_team} average goals away: "
      f"{team_goal_averages.loc[away_team, 'avg_goals_away']}")
```

```python
all_teams = pd.Index(
    pd.concat([df["HomeTeam"], df["AwayTeam"]]).unique()
)

all_team_wins = (
    wins_by_team
    .reindex(all_teams, fill_value=0)
    .sort_values(ascending=False)
)

print(all_team_wins.to_string())
```
